# data exploration

```python
import os
import json


DATA_DIR = "../data"
with open(os.path.join(DATA_DIR, "train.jsonl")) as f:                                                                       
    train_data = [json.loads(line) for line in f]                                                                            
                                                                                                                               
print(f"Training samples: {len(train_data)}")
```

```python
# Look at actual memes with their labels
from PIL import Image as PILImage
from IPython.display import display

for sample in train_data[:5]:
    img_path = os.path.join(DATA_DIR, sample['img'])
    img = PILImage.open(img_path)
    
    label = "HATEFUL" if sample['label'] == 1 else "NOT HATEFUL"
    print(f"\n--- {label} ---")
    print(f"Text: {sample['text']}")
    display(img)
```

```python
# Check class balance — important for pos_weight in loss function later
labels = [s['label'] for s in train_data]
hateful = sum(labels)
not_hateful = len(labels) - hateful

print(f"Total: {len(labels)}")
print(f"Hateful: {hateful} ({100*hateful/len(labels):.1f}%)")
print(f"Not hateful: {not_hateful} ({100*not_hateful/len(labels):.1f}%)")
print(f"\npos_weight for loss = {not_hateful/hateful:.2f}")
```

```python
# Look at some random HATEFUL memes specifically — build intuition
import random
random.seed(42)

hateful_samples = [s for s in train_data if s['label'] == 1]
print(f"Total hateful memes: {len(hateful_samples)}")

for sample in random.sample(hateful_samples, 4):
    img_path = os.path.join(DATA_DIR, sample['img'])
    img = PILImage.open(img_path)
    print(f"\n--- HATEFUL ---")
    print(f"Text: {sample['text']}")
    display(img)
```

```python
# Also load dev (validation) and test sets — check their sizes
with open(os.path.join(DATA_DIR, "dev.jsonl")) as f:
    dev_data = [json.loads(line) for line in f]

with open(os.path.join(DATA_DIR, "test.jsonl")) as f:
    test_data = [json.loads(line) for line in f]

print(f"Train: {len(train_data)} samples")
print(f"Dev:   {len(dev_data)} samples")
print(f"Test:  {len(test_data)} samples")
print(f"Total: {len(train_data) + len(dev_data) + len(test_data)} samples")
```
